```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs

# Load the dataset
ds = xr.open_dataset(paths,decode_times=False).isel(time=slice(1, None)).mean(dim='time')
ds_exp = xr.open_dataset(paths_exp,decode_times=False).isel(time=slice(1, None)).mean(dim='time')
 # Constants
p0 = 100000  # reference pressure in Pa

# Calculating pressure from hybrid level coefficients
pressure = ds['a'] + ds['b'] * ds['ps']
pressure_zonal_mean = pressure.mean(dim='lon')
# Calculate the zonal mean (mean over longitude)
cli_zonal_mean = ds_exp['cli'].mean(dim='lon') - ds['cli'].mean(dim='lon').values

# Plotting
fig, ax = plt.subplots(figsize=(10, 6))

# Plot a filled contour on the axes object
contour = ax.contourf(ds['lat'], pressure_zonal_mean.mean(dim='lat'),
                      cli_zonal_mean, cmap='bwr', vmin=-5e-6, vmax=5e-6)

# Invert y-axis for pressure
ax.invert_yaxis()

# Switch y-axis to logarithmic scale
ax.set_yscale('log')

# Add labels, colorbar, and title using the axes object
cbar = fig.colorbar(contour, ax=ax, label='Mass fraction Cloud Ice')
ax.set_xlabel('Latitude')
ax.set_ylabel('Pressure (Pa)')
ax.set_title('Zonal Mean Contour Plot of Cloud Liquid Water Content')
ax.set_ylim(100000, 10000)
# Show plot
plt.show()
```

```python
import xarray as xr
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
from workflow.scripts.utils import calc_error, calc_error_gridded, regrid_global
from workflow.scripts.plotting_tools import global_map, create_facet_plot
from pyclim_noresm.general_util_funcs import global_avg
import numpy as np
import matplotlib as mpl
import pandas as pd
```

```python
paths = sorted(snakemake.input.paths)
vname = snakemake.wildcards.vName

params = snakemake.params
time_slice = params.get('time_slice', slice(3,-1))
nlevels = params.get('nlevels', 11)
draw_error_mask = params.get('draw_error_mask', True)
```

```python
erfs = {}
erfs['mean'] = {}
erfs['std'] = {}
for path in paths:
    ds = xr.open_dataset(path)
    if 'year' in ds.dims:
        ds = ds.rename_dims(year='time')
    ds = ds.isel(time=time_slice)

    source = ds.source_id
    ts = global_avg(ds[vname])
    ts = ts.mean(dim='time')
    std = calc_error(ds[vname], kind='SEM',time_dim='time')
    erfs['mean'][source] = ts.values
    erfs['std'][source] = std.values
    
```

```python
df = pd.DataFrame(erfs)
nmodels = len(df['mean'])
multimodel_mean= df['mean'].mean()
error = df['mean'].std()/np.sqrt(nmodels)
df.loc['Multi-model','mean'] = multimodel_mean 
df.loc['Multi-model', 'std'] =error
vmax=round(abs(df.loc[:,'mean']).max()+0.1,1)
df = df.astype(float)
```

```python
dsets = {}
for path in paths:

    ds = xr.open_dataset(path)
    source_id = ds.source_id
    print(source_id)
    with xr.set_options(keep_attrs=True):
        
        if snakemake.config.get('regrid_params', None) and snakemake.params.get('regrid', False):
            if not ds.cf.bounds.get('lon',None):
                ds = ds.cf.add_bounds(['lon','lat'])
        
            grid_params = snakemake.config['regrid_params']
            grid_path = grid_params.get('grid_path',None)
            method=grid_params.get('method','conservative')
            if grid_path:
                out_grid = xr.open_dataset(grid_path)
                ds = regrid_global(ds, out_grid, method=method)
            elif grid_params.get('dxdy',None):
                dxdy = grid_params['dxdy']
                ds = regrid_global(ds, lon=dxdy[0], lat=dxdy[1], method=method)
            else:
                print('No outgrid provided!')


    dsets[source_id]={}
    ds = ds[snakemake.wildcards.vName].isel(year=time_slice)
    error = calc_error_gridded(ds,kind='SEM' )
    ds = ds.mean(dim='year',keep_attrs=True).to_dataset(name=f'{source_id}')
    dsets[source_id]['data']=ds
    dsets[source_id]['error']=error
if vname in ['LWDirectEff','LWDirectEff_cs']:
    cvmin=-2
    cvmax=2
else:
    cvmin=-5
    cvmax=5
     
```

```python
def calc_multi_model_mean(dsets):
    temp_ds = []
    n = len(dsets)
    k = list(dsets.keys())[0]
    for ds in dsets:
        temp_ds.append(dsets[k]['data'][k])
    dsets = xr.merge(temp_ds) 
    dsets['multi model mean'] = xr.zeros_like(dsets[k])
    for dvar in dsets.data_vars:
        dsets['multi model mean'] += dsets[dvar]
    
    dset = dsets['multi model mean']/n
    dset = dset.to_dataset(name ='multi model mean')
#     total = total.to_dataset(name='Multi model mean')
    return dset
```

```python
if snakemake.params.get('add_multimodel_mean', False):
    dsets['multi model mean'] = {}
    dsets['multi model mean']['data'] = calc_multi_model_mean(dsets)
```

```python
fig,ax,cax = create_facet_plot(len(dsets),
                            subplot_kw={'projection':ccrs.EckertIV()}, 
                figsize=(8,4), last_axis_plain=False,create_cax=False)
fig.set_dpi(300)
cmap = mpl.cm.get_cmap('RdYlBu_r', nlevels)


for k, i in zip(list(ax.keys()), dsets.keys()):
    mesh=dsets[i]['data'][i].plot.pcolormesh(ax=ax[k],vmin=cvmin,
                              vmax=cvmax,cmap=cmap, 
                                   add_colorbar=False, transform=ccrs.PlateCarree())

    error_mask = dsets[i]['data'][i] < dsets[i].get('error',10000.0)
    if draw_error_mask:
        error_mask.plot.contourf(ax=ax[k],hatches=[None,'...'],alpha=0., levels=3, add_colorbar=False)
    ax[k].coastlines(linewidth=.8)
    ax[k].set_title(i, fontsize=7, pad=0.3)
    # global_map(ax[k])
    globm = global_avg(dsets[i]['data'][i]).values
    ax[k].text(0.54, 0.135, f"{globm:.2f} $\mathrm{{W\; m}}^{{-2}}$", ha='center', va='center', transform=ax[k].transAxes, zorder=1230,
        bbox=dict(facecolor='white', edgecolor='black', pad=0.2, alpha=.8, boxstyle='round,pad=1'), fontsize=6)
cax = fig.add_axes([0.25,0.02,0.5,0.04])
cmap = mpl.cm.RdYlBu_r
# last = list(ax.keys())[-1]
# df['mean'].plot.barh(ax=ax[last], legend=False, xerr=df['std'])
# ax[last].axvline(color='darkgrey')
# ax[last].set_xlim(-vmax,vmax)
# ax[last].set_xlabel('$W/m^2$')
# ax[last].yaxis.tick_right()
cbar=fig.colorbar(mesh, cax=cax, extend='both' 
             , orientation='horizontal')
cbar.ax.tick_params(labelsize=6)
cbar.set_label(label='$W/m^2$',fontsize=7)
plt.savefig(snakemake.output.outpath, bbox_inches='tight', facecolor='white', dpi=300)
```

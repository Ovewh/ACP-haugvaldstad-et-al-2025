```python
def calc_cld_lev(ds_cl,height_def, vname='cl', ps=None, ds_pfull=None):
    ds_cl.attrs['variable_id'] = vname
    if ps is not None:
        ds_cl = ds_cl.assign_coords(ps=ps)
    if ds_cl.source_id == 'IPSL-CM6A-LR-INCA':
        rg_data = ds_cl.rename(lev='plev')
    elif ds_cl.source_id == 'UKESM1-0-LL':
        pass

    else:
        rg_data = model_levels_to_pressure_levels(ds_cl).compute()
    
    for h, hdef in height_def.items():
        if ds_cl.source_id == 'UKESM1-0-LL':
            print(h)
            pmask = xr.where((hdef['bottom']*100 > ds_pfull) &  (ds_pfull > hdef['top']*100),True, False) 
            temp_rg_data = ds_cl.where(pmask).max(dim='lev')
        else:
            if hdef['top']:
                t = hdef['top']*100
            if hdef['bottom']:
                b = hdef['bottom']*100
            temp_rg_data = rg_data.sel(plev=slice(b, t)).max(dim='plev')
        temp_rg_data[vname].attrs['comment'] = f'Percentage {h} cloud cover, including both large-scale and convective cloud.'
        temp_rg_data = temp_rg_data.rename({vname:f'{vname}_{h}'})
        out_ds = temp_rg_data
    return out_ds    
```

```python
def model_levels_to_pressure_levels(ds: xr.Dataset | xr.DataArray, 
                                    plevels=np.array([100000.,  92500.,  85000.,  70000.,  60000.,  50000.,  40000.,  30000.,
        25000.,  20000.,  15000.,  10000.,   7000.,   5000.,   3000.,   2000.,
         1000.,    500.,    100.])):
    """
    Convert model levels to pressure levels.
    """
    try:
        import geocat.comp as geocomp
    except ImportError:
        raise ImportError(
            "geocat-comp is required for this function. Run snakemake with --use-conda to enable it."
        )
    ds = ds.copy()
    if "ap" in ds.data_vars:
        ds = ds.rename_vars({"ap": "a"})

    if "lev_bounds" in ds.data_vars:
        ds = ds.rename_vars({"lev_bounds": "lev_bnds"})

    
    da = ds
    if ds.cf["Z"].formula == "p = a*p0 + b*ps":
        if "a" in da.coords:
            da = da.drop("a")
        if "b" in da.coords:
            da = da.drop("b")
        if "ps" in da.coords:
            da = da.drop("ps")
        da = geocomp.interpolation.interp_hybrid_to_pressure(
            data=da,
            ps=ds["ps"],
            hyam=ds["a"],
            hybm=ds["b"],
            p0=ds.get("p0", 100000.0),
            new_levels=plevels
        )
    
    elif ds.cf["Z"].formula in ["p = ap + b*ps", "p(n,k,j,i) = ap(k) + b(k)*ps(n,j,i)"]:
        formula_terms = da.cf.formula_terms

        hyam = ds["a"]
        a = "a"

        if a in da.coords:
            da = da.drop(a)
        if "b" in da.coords:
            da = da.drop("b")
        if "ps" in da.coords:
            da = da.drop("ps")

        da = geocomp.interpolation.interp_hybrid_to_pressure(
            data=da[ds.variable_id], ps=ds["ps"], hyam=hyam, hybm=ds["b"], p0=ds.get("p0", 1),
            new_levels=plevels, lev_dim="lev"
        )

    else:
        print("formula not reconized")

    outds = da.to_dataset(name=ds.variable_id)
    outds.attrs = ds.attrs

    outds.attrs["history"] = (
        outds.attrs.get("history", "")
        + f"@{time.ctime()} converted to standard pressure levels"
    )
    return outds
```

```python
cld_def = {snakemake.wildcards.height:snakemake.config['cloud_def'].get(snakemake.wildcards.height)}
```

```python
cl = xr.open_dataset(snakemake.input.cl[0])

clfra = calc_cld_lev(cl,cld_def, 
                           ps = cl.get('ps', None),
                          ds_pfull=cl.get('pfull', None)
                      )
```

```python
clfra.attrs = cl.attrs

clfra.attrs['variable_id'] = f'cl_{snakemake.wildcards.height}'

clfra.to_netcdf(snakemake.output.outpath)
```
